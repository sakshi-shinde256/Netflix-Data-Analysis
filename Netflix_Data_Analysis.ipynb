```python
df.drop_duplicates(inplace=True)

df.duplicated().sum()
```

```python
df.fillna("Unknown", inplace=True)

df.isnull().sum()
```

```python
content_type = df['type'].value_counts()

print(content_type)
```

```python
content_type.plot(kind='bar')

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

print(top_countries)
```

```python
top_countries.sort_values().plot(kind='barh')

plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
yearly_content = df['release_year'].value_counts().sort_index()

print(yearly_content.tail(10))
```

```python
yearly_content.plot(kind='line', figsize=(10,5))

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

```python
ratings = df['rating'].value_counts().head(10)

print(ratings)
```

```python
ratings.plot(kind='bar')

plt.title("Top Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
top_directors = df['director'].value_counts().head(10)

print(top_directors)
```

```python
top_directors.sort_values().plot(kind='barh')

plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()
```

```python
movies = df[df['type'] == 'Movie']
tv_shows = df[df['type'] == 'TV Show']

print("Total Movies:", len(movies))
print("Total TV Shows:", len(tv_shows))
```

```python
movies['duration'].value_counts().head(10)
```

```python
genres = df['listed_in'].value_counts().head(10)

print(genres)
```

```python
genres.sort_values().plot(kind='barh')

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

```python
print("Netflix Data Analysis Summary")
print("--------------------------------")
print("Total Titles:", len(df))
print("Total Movies:", len(movies))
print("Total TV Shows:", len(tv_shows))
print("Top Country:", df['country'].value_counts().index[0])
print("Most Common Rating:", df['rating'].value_counts().index[0])
```
